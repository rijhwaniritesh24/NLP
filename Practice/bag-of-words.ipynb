```python
import pandas as pd
import numpy as np
import os
spam_message= pd.read_csv('../lectures/spam.csv',encoding='latin-1')
```

```python
message = spam_message.iloc[:,:-3]
message= message.rename(columns={'v1': 'label', 'v2':'message'})
```

```python
message
```

```python
## Data cleaning and stopwords
import re
import nltk
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
stem = PorterStemmer()
```

```python
corpus =[]
for i in range(len(message)):
    review = re.sub('[^a-zA-z]',' ', message['message'][i])
    review.lower()
    review = review.split()
    review = [ stem.stem(word) for word in review if not word in stopwords.words('english')]
    review = ' '.join(review)
    corpus.append(review)
    
```

```python
corpus
```

```python
## Create a bag of words model
from sklearn.feature_extraction.text import CountVectorizer

bag_of_words = CountVectorizer(max_features=2500, binary=True)
```

```python
X = bag_of_words.fit_transform(corpus).toarray()
```
